Задание 1.4

Напишите программу, которая считывает число (время в секундах) выводит его в формате: часы, минуты и секунды.

Для вычислений рекомендуем использовать операцию целочисленного деления `//` и взятия остатка от деления `%`

```python
class TimeConverter:
    """Класс для работы с преобразованием времени"""

    SECONDS_IN_MINUTE = 60
    SECONDS_IN_HOUR = 3600
    SECONDS_IN_DAY = 86400

    def __init__(self, total_seconds=0):
        self.total_seconds = total_seconds
        self._calculate_components()

    def _calculate_components(self):
        """Вычисляет компоненты времени"""
        sign = -1 if self.total_seconds < 0 else 1
        abs_seconds = abs(self.total_seconds)

        self.days = abs_seconds // self.SECONDS_IN_DAY
        remaining = abs_seconds % self.SECONDS_IN_DAY

        self.hours = remaining // self.SECONDS_IN_HOUR
        remaining %= self.SECONDS_IN_HOUR

        self.minutes = remaining // self.SECONDS_IN_MINUTE
        self.seconds = remaining % self.SECONDS_IN_MINUTE

        # Применяем знак только к наибольшему ненулевому компоненту
        if sign == -1:
            if self.days > 0:
                self.days = -self.days
            elif self.hours > 0:
                self.hours = -self.hours
            elif self.minutes > 0:
                self.minutes = -self.minutes
            else:
                self.seconds = -self.seconds

    def format_short(self):
        """Формат ЧЧ:ММ:СС"""
        sign = '-' if self.total_seconds < 0 else ''
        return f"{sign}{abs(self.hours + self.days * 24):02d}:{abs(self.minutes):02d}:{abs(self.seconds):02d}"

    def format_long(self):
        """Развернутый формат с днями"""
        parts = []
        if abs(self.days) > 0:
            parts.append(f"{self.days} дн.")
        if abs(self.hours) > 0 or self.days != 0:
            parts.append(f"{abs(self.hours)} час.")
        if abs(self.minutes) > 0 or self.hours != 0 or self.days != 0:
            parts.append(f"{abs(self.minutes)} мин.")
        parts.append(f"{abs(self.seconds)} сек.")

        # Добавляем знак если время отрицательное
        prefix = '-' if self.total_seconds < 0 else ''
        return prefix + ' '.join(parts)

    def format_human(self):
        """Человекочитаемый формат"""
        components = []

        if abs(self.days) > 0:
            day_word = self._pluralize(abs(self.days), ["день", "дня", "дней"])
            components.append(f"{abs(self.days)} {day_word}")

        if abs(self.hours) > 0:
            hour_word = self._pluralize(abs(self.hours), ["час", "часа", "часов"])
            components.append(f"{abs(self.hours)} {hour_word}")

        if abs(self.minutes) > 0:
            minute_word = self._pluralize(abs(self.minutes), ["минута", "минуты", "минут"])
            components.append(f"{abs(self.minutes)} {minute_word}")

        if abs(self.seconds) > 0 or not components:
            second_word = self._pluralize(abs(self.seconds), ["секунда", "секунды", "секунд"])
            components.append(f"{abs(self.seconds)} {second_word}")

        result = ', '.join(components)
        if self.total_seconds < 0:
            result = f"минус {result}"

        return result

    def _pluralize(self, number, forms):
        """Правильное склонение для русских числительных"""
        if number % 10 == 1 and number % 100 != 11:
            return forms[0]
        elif 2 <= number % 10 <= 4 and (number % 100 < 10 or number % 100 >= 20):
            return forms[1]
        else:
            return forms[2]

    def to_timestamp(self):
        """Возвращает строку времени (если в пределах суток)"""
        if abs(self.total_seconds) >= self.SECONDS_IN_DAY:
            return "Более суток"

        total = self.total_seconds % self.SECONDS_IN_DAY
        if total < 0:
            total += self.SECONDS_IN_DAY

        hours = total // self.SECONDS_IN_HOUR
        minutes = (total % self.SECONDS_IN_HOUR) // self.SECONDS_IN_MINUTE
        seconds = total % self.SECONDS_IN_MINUTE

        return f"{hours:02d}:{minutes:02d}:{seconds:02d}"


def parse_time_input(time_str):
    """Парсинг ввода в разных форматах"""
    time_str = time_str.strip().lower()

    # Проверяем, является ли ввод просто числом (секунды)
    if time_str.replace('-', '').isdigit():
        return int(time_str)

    # Пробуем парсинг формата ЧЧ:ММ:СС
    if ':' in time_str:
        parts = time_str.split(':')
        if len(parts) == 3 and all(p.isdigit() for p in parts):
            hours, minutes, seconds = map(int, parts)
            return hours * 3600 + minutes * 60 + seconds

    # Пробуем парсинг формата с суффиксами
    multipliers = {'ч': 3600, 'м': 60, 'с': 1, 'д': 86400}
    total = 0

    import re
    pattern = r'(-?\d+)\s*([чмсд])'
    matches = re.findall(pattern, time_str)

    if matches:
        for value, unit in matches:
            total += int(value) * multipliers[unit]
        return total

    raise ValueError("Неверный формат времени")


def main():
    """Основная функция программы"""
    print("=== Расширенный конвертер времени ===")
    print("Поддерживаемые форматы ввода:")
    print("  1. Секунды (например: 3665)")
    print("  2. ЧЧ:ММ:СС (например: 1:01:05)")
    print("  3. С суффиксами (например: 1ч 30м 25с)")
    print("  4. Отрицательные значения (например: -120, -0:2:0)")
    print("-" * 40)

    while True:
        try:
            user_input = input("\nВведите время (или 'выход' для завершения): ")

            if user_input.lower() in ['выход', 'exit', 'quit', 'q']:
                print("Программа завершена.")
                break

            if not user_input.strip():
                print("Ввод не может быть пустым!")
                continue

            # Преобразуем ввод в секунды
            total_seconds = parse_time_input(user_input)

            # Создаем объект TimeConverter
            converter = TimeConverter(total_seconds)

            # Выводим результаты
            print("\n" + "=" * 50)
            print(f"Введенное значение: {user_input}")
            print(f"Всего секунд: {total_seconds}")
            print(f"Короткий формат: {converter.format_short()}")
            print(f"Расширенный формат: {converter.format_long()}")
            print(f"Человекочитаемый формат: {converter.format_human()}")
            print(f"Временная метка: {converter.to_timestamp()}")

            # Дополнительная информация
            if abs(total_seconds) >= converter.SECONDS_IN_DAY:
                days = abs(total_seconds) / converter.SECONDS_IN_DAY
                print(f"\nИнтересный факт: это примерно {days:.2f} суток")

            print("=" * 50)

        except ValueError as e:
            print(f"Ошибка: {e}")
            print("Пожалуйста, введите время в одном из поддерживаемых форматов.")
        except KeyboardInterrupt:
            print("\n\nПрограмма прервана пользователем.")
            break
        except Exception as e:
            print(f"Неожиданная ошибка: {e}")


def run_tests():
    """Функция для тестирования"""
    test_cases = [
        0, 1, 59, 60, 61, 3599, 3600, 3601,
        3665, 86400, 90061, 172800,
        -1, -60, -3665, -86400
    ]

    print("Тестирование конвертера времени:")
    print("-" * 60)

    for seconds in test_cases:
        converter = TimeConverter(seconds)
        print(f"{seconds:8d} сек => {converter.format_short():12s} | {converter.format_human()}")

    print("-" * 60)

    # Тест парсинга
    test_inputs = ["3665", "1:01:05", "1ч 1м 5с", "-2:30:00"]
    for inp in test_inputs:
        try:
            result = parse_time_input(inp)
            print(f"Парсинг '{inp}' => {result} секунд")
        except ValueError as e:
            print(f"Ошибка парсинга '{inp}': {e}")


if __name__ == "__main__":
    # Запуск тестов при непосредственном запуске файла
    run_tests()
    print("\n\n")

    # Запуск основной программы
    main()
```

[GC-состав](https://ru.wikipedia.org/wiki/GC-%D1%81%D0%BE%D1%81%D1%82%D0%B0%D0%B2) является важной характеристикой геномных последовательностей и определяется как процентное соотношение суммы всех гуанинов и цитозинов к общему числу нуклеиновых оснований в геномной последовательности.

Напишите программу, которая вычисляет процентное содержание символов G (гуанин) и C (цитозин) в введенной строке (программа **не должна зависеть** от регистра вводимых символов).

Например, в строке "acggtgttat" процентное содержание символов G и C равно $\dfrac{4}{10}\cdot 100 = 40.0$, где 4 - это количество символов G и C,  а 10 - это длина строки.

```python

dna = input()
dna_upper = dna.upper()
gc_count = dna_upper.count('G') + dna_upper.count('C')
total_length = len(dna)
gc_percentage = (gc_count / total_length) * 100
print(gc_percentage)
```

Напишите программу, которая выводит наибольний и наименьший элементсписка (можно для упрощения создать список внутри программы, без считывания). Решение не должно использовать встроенную функцию `max()` или `min()`.

```python
numbers = [12, 45, 2, 41, 31, 10, 8, 6, 4]
me = numbers[0]
maxe = numbers[0]
for num in numbers:
    if num < me:
        me = num
    if num > maxe:
        maxe = num
print(me)
print(maxe)
```
